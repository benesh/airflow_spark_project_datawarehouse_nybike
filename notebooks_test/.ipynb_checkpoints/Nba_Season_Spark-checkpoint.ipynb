```python
### Gold LAYER 

spark.sql(f"""
CREATE TABLE {catalog_name}.{gold_name_database}.{table_name_gold_fact_trip}(
    trip_uuid string ,
	enr_rideable_type string ,
    start_at timestamp ,
	stop_at timestamp ,
    enr_trip_duration double  
)
USING iceberg
PARTITIONED BY(year(start_at))
;
""")


spark.sql(f"""
CREATE TABLE {catalog_name}.{gold_name_database}.{table_name_gold_dim_location}(
    trip_uuid string,
	start_station_id string,
	start_station_name string,
	start_station_latitude string,
	start_station_longitude string,
	end_station_id string,
	end_station_name string,
	end_station_latitude string,
	end_station_longitude string,
    start_at timestamp
)
USING iceberg
PARTITIONED BY(year(start_at))
;
    """)


spark.sql(f"""
CREATE TABLE {catalog_name}.{gold_name_database}.{table_name_gold_dim_customer}(
    trip_uuid string,
	user_type string,
    enr_user_type string,
    gender integer,
    enr_gender string,
	customer_year_birth string
)
USING iceberg
PARTITIONED BY(enr_user_type,enr_gender)
;
    """)


spark.sql(f"""
CREATE TABLE {catalog_name}.{gold_name_database}.{table_name_gold_dim_times}(
    trip_uuid string,
    enr_year integer,
    enr_quarter integer,
    enr_quarter_name string,
    enr_month integer,
    enr_month_name string,
    enr_day integer,
    enr_weekday integer,
    enr_weekday_name string,
    start_at timestamp,
	stop_at timestamp
)
USING iceberg
PARTITIONED BY(enr_year,enr_quarter,enr_month,bucket(10,enr_day))
;
    """)
```
